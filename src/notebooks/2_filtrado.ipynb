# 2. Filtrado/tratamiento de *Null*

```python
alimentacion["nutriscore_grade"].isna().sum()
```

```python
alimentacion["nutriscore_grade"].unique()
```

```python
alimentacion["ingredients_text"] = alimentacion["ingredients_text"].fillna("")
alimentacion["ingredients_analysis_tags"] = alimentacion["ingredients_analysis_tags"].fillna("Sin informarción")  # Asignamos "" a los productos sin análisis de ingredientes definido
alimentacion["food_groups"] = alimentacion["food_groups"].fillna("Sin categoría")  # Asignamos "" a los productos sin grupo de alimentos definido
alimentacion["nova_group"] = alimentacion["nova_group"].fillna(0)  # Asignamos "0" a los productos sin grupo NOVA definido
alimentacion["nutriscore_grade"] = alimentacion["nutriscore_grade"].fillna("unknown")  # Asignamos "unknown" a los productos sin Nutri-Score definido
alimentacion["environmental_score_grade"] = alimentacion["environmental_score_grade"].fillna("unknown")  # Asignamos "unknown" a los productos sin puntaje ambiental definido

alimentacion["stores"] = alimentacion["stores"].fillna("Sin información")  # Asignamos "Sin información" a los productos sin tienda definida
alimentacion["brands"] = alimentacion["brands"].fillna("Sin información")  # Asignamos "Sin información" a los productos sin marca definida
alimentacion["categories_en"] = alimentacion["categories_en"].fillna("Sin información")  # Asignamos "Sin información" a los productos sin categoría definida
alimentacion["labels_en"] = alimentacion["labels_en"].fillna("Sin información")  # Asignamos "Sin información" a los productos sin etiqueta definida
alimentacion["food_groups_en"] = alimentacion["food_groups_en"].fillna("Sin información")  # Asignamos "Sin información" a los productos sin grupo de alimentos definido
alimentacion["additives_en"] = alimentacion["additives_en"].fillna("Sin información")  # Asignamos "Sin información" a los productos sin aditivo definido
alimentacion["additives_tags"] = alimentacion["additives_tags"].fillna("Sin información")  # Asignamos "Sin información" a los productos sin aditivo definido
```

```python
alimentacion.info()
```

```python
print("Cantidad productos iniciales:", len(alimentacion["categories_en"]))
alimentacion = alimentacion[alimentacion["categories_en"] != "Non food products"]
print("Cantidad productos después de filtrado por 'Non food products':", len(alimentacion["categories_en"]))
```

```python
# Lista de palabras clave que indican productos NO alimentarios (Inglés y Español)
# Incluye raíces para capturar variaciones (ej: "shampoo" captura "shampoos", "en:organic-shampoo")
palabras_no_alimentacion = [
    # --- Higiene y Cuidado Personal ---
    "soap", "jabón", "jabon", "shampoo", "champú", "champú", "conditioner", "acondicionador",
    "gel de baño", "gel hidroalcohólico", "gel higienizante",
    "deodorant", "desodorante", "antiperspirant", "antitranspirante",
    "toothpaste", "pasta de dientes", "dentífrico",
    "mouthwash", "enjuague", "dental.*floss", "hilo dental",
    
    # --- Cosmética y Maquillaje ---
    "cosmetic", "cosmética", "cosmetica", "makeup", "maquillaje", "beauty",
    "lipstick", "pintalabios", "labial",
    "mascara", "rímel", "rimel",
    "eye-pencil", "eye-liner", "delineador", "sombra de ojos",
    "foundation", "base de maquillaje", "corrector",
    "perfume", "cologne", "colonia", "fragrance", "fragancia",
    "aftershave", "after sun", "serums", "face"
    
    # --- Cuidado de la piel y cabello ---
    "sunscreen", "suncare", "sun.*care", "protector solar", "bronceador",
    "skincare", "skin.*care", "cuidado de la piel", "hygiene",
    "haircare", "hair.*care", "cuidado del cabello", "tinte.*pelo", "tinte.*capilar",
    "shaving", "razor", "cuchilla", "maquinilla", "espuma.*afeitar",
    
    # --- Limpieza del hogar (no comestible) ---
    "detergent", "detergente", "bleach", "lejía", "lejia", "cleaner", "limpiador",
    "dishwasher", "lavavajillas", "laundry", "jabón.*ropa", "suavizante",
    "ambientador", "air.*freshener",
    "toallitas húmedas", "toallitas desmaquillantes",
    
    # --- Palabras "radicales" (Si aparecen en cualquier lado, borra el producto) ---
    "body", "hand", "crema", "lotion", "loción", "software", "cat", "mascotas", 
    "tobacco", "creams", "media", "accessories", "crayons", "parfum"
]

# Creamos un patrón regex para buscar cualquiera de estas palabras
patron_no_comida = '|'.join(palabras_no_alimentacion)

# Filtramos el DataFrame: nos quedamos SOLO con las filas que NO contengan esas palabras
# Buscamos en categories_tags y en product_name por seguridad
# regex=True es implícito al pasar un patrón con '|', pero lo explicitamos por claridad
mask_alimentacion = ~(
    alimentacion['categories_tags'].str.contains(patron_no_comida, case=False, na=False, regex=True) |
    alimentacion['product_name'].str.contains(patron_no_comida, case=False, na=False, regex=True)
)

alimentacion_limpio = alimentacion[mask_alimentacion].copy()
alimentacion_limpio = alimentacion_limpio.reset_index(drop=True)

print(f"✅ Filtrado completado:")
print(f"   - Antes: {len(alimentacion)} filas")
print(f"   - Después: {len(alimentacion_limpio)} filas")
print(f"   - Eliminadas: {len(alimentacion) - len(alimentacion_limpio)} filas de cosmética/higiene")
```
